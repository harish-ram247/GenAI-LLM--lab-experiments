```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch


# ============================================================
# 1. LOAD GPT-2 MODEL
# ============================================================

model_name = "gpt2"

print("Loading tokenizer...")

tokenizer = GPT2Tokenizer.from_pretrained(model_name)

print("Loading model...")

model = GPT2LMHeadModel.from_pretrained(model_name)

print("Model loaded successfully!")


# GPT-2 does not have a padding token by default
tokenizer.pad_token = tokenizer.eos_token


# ============================================================
# 2. CODE GENERATION FUNCTION
# ============================================================

def generate_code(prompt, max_new_tokens=80):

    # Convert prompt into tokens
    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    # Generate output
    with torch.no_grad():

        output = model.generate(
            input_ids=inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            max_new_tokens=max_new_tokens,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=False
        )

    # Convert tokens back to text
    generated_text = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return generated_text


# ============================================================
# 3. CODE GENERATION FROM NATURAL LANGUAGE
# ============================================================

prompt1 = """# Write a Python function to check if a number is prime

def is_prime(n):
"""

print("\n================================")
print("GENERATED FUNCTION")
print("================================")

result1 = generate_code(
    prompt1,
    max_new_tokens=80
)

print(result1)


# ============================================================
# 4. DEBUGGING A FAULTY CODE SNIPPET
# ============================================================

buggy_code = """# The following function should return the factorial of n,
# but it has a bug. Fix it.

def factorial(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result

# Corrected function:

def factorial_fixed(n):
"""

print("\n================================")
print("DEBUG SUGGESTION")
print("================================")

result2 = generate_code(
    buggy_code,
    max_new_tokens=80
)

print(result2)
```
